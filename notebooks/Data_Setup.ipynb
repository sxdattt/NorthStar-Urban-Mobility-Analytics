```python
GITHUB_RAW = "https://raw.githubusercontent.com/sxdattt/NorthStar-Urban-Mobility-Analytics/main/"

files = {
    "orders":       "orders.csv",
    "deliveries":   "deliveries.csv",
    "customers":    "customers.csv",
    "complaints":   "complaints.csv",
    "drivers":      "drivers.csv",
    "vehicles":     "vehicles.csv",
    "hubs":         "hubs.csv",
    "incidents":    "incidents.csv",
    "app_events":   "app_events.csv",
    "data_dict":    "data_dictionary.csv"
}

print("✅ URLs configured. Sample check:")
print(GITHUB_RAW + files["hubs"])
```

```python
import pandas as pd
import numpy as np
import requests
from io import StringIO

pd.set_option('display.max_columns', None)
pd.set_option('display.float_format', '{:.2f}'.format)

print("Libraries imported successfully ✅")
```

```python
print("URLs being attempted:\n")
for name, filename in files.items():
    url = GITHUB_RAW + filename
    print(f"{name}: {url}")
```

```python
# Load all CSVs from GitHub into a dictionary of DataFrames
dataframes = {}

for name, filename in files.items():
    url = GITHUB_RAW + filename
    response = requests.get(url)

    if response.status_code == 200:
        dataframes[name] = pd.read_csv(StringIO(response.text))
        print(f"✅ Loaded '{name}' — {dataframes[name].shape[0]} rows × {dataframes[name].shape[1]} cols")
    else:
        print(f"❌ FAILED to load '{name}' — Check filename or GitHub URL (Status: {response.status_code})")
```

```python
orders      = dataframes["orders"]
deliveries  = dataframes["deliveries"]
customers   = dataframes["customers"]
complaints  = dataframes["complaints"]
drivers     = dataframes["drivers"]
vehicles    = dataframes["vehicles"]
hubs        = dataframes["hubs"]
incidents   = dataframes["incidents"]
app_events  = dataframes["app_events"]
data_dict   = dataframes["data_dict"]

print("All datasets assigned to variables ✅")
```

```python
print("=" * 60)
print(f"{'DATASET':<15} {'ROWS':>8} {'COLUMNS':>10}")
print("=" * 60)

for name, df in dataframes.items():
    print(f"{name:<15} {df.shape[0]:>8,} {df.shape[1]:>10}")

print("=" * 60)
print(f"{'TOTAL FILES':<15} {len(dataframes):>8}")
```

```python
for name, df in dataframes.items():
    print(f"\n{'='*55}")
    print(f"📋  {name.upper()}")
    print(f"{'='*55}")
    display(df.head(3))
```

```python
print("MISSING VALUES REPORT")
print("=" * 55)

for name, df in dataframes.items():
    missing = df.isna().sum()
    missing_cols = missing[missing > 0]

    if len(missing_cols) > 0:
        print(f"\n⚠️  {name.upper()} — {len(missing_cols)} column(s) with missing values:")
        for col, count in missing_cols.items():
            pct = (count / len(df)) * 100
            print(f"     • {col}: {count} missing ({pct:.1f}%)")
    else:
        print(f"✅  {name.upper()} — No missing values")
```

```python
for name, df in dataframes.items():
    print(f"\n{'='*55}")
    print(f"📋  {name.upper()} — Column Types")
    print(f"{'='*55}")
    print(df.dtypes.to_string())
```

```python
join_checks = {
    "orders → customers":  ("orders",    "customer_id",  "customers",  "customer_id"),
    "orders → deliveries": ("orders",    "order_id",     "deliveries", "order_id"),
    "orders → complaints": ("orders",    "order_id",     "complaints", "order_id"),
    "drivers → vehicles":  ("drivers",   "driver_id",    "incidents",  "driver_id"),
}

print("JOIN KEY VALIDATION")
print("=" * 55)

for join_name, (t1, col1, t2, col2) in join_checks.items():
    ok1 = col1 in dataframes[t1].columns
    ok2 = col2 in dataframes[t2].columns
    status = "✅" if ok1 and ok2 else "❌"
    print(f"{status}  {join_name}: [{col1}] ↔ [{col2}]")
```
